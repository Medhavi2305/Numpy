## Numpy

```python
import sys
import numpy as np
import time

print(sys.executable)
print(np.__version__)

arr = np.array([1, 2, 3])
print(arr)
```

```python
arr_1d = np.array([1, 2, 3])
print('1D Array: \n',arr_1d)

arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print('2D Array: \n', arr_2d)
```

```python
py_list = [1, 2, 3]
print('Python List Multiplication: \n', py_list * 2)

numpy_arr = np.array([1, 2, 3])
print('Python Array Multiplication: \n', numpy_arr * 2)
```

```python
start = time.time()
py_list = [i*2 for i in range(1000)]
print('List Operation Time: ', time.time() - start)

start = time.time()
py_Arr = np.arange(1000) * 2
print('Array Operation Time: ', time.time() - start)
```

## Creating Array From Scratch

```python
zeros = np.zeros((3, 4))
print('Zeros Array: \n', zeros)

ones = np.ones((2, 3))
print('Ones Array: \n', ones)

constant_Array = np.full((2, 3), 7)
print('Constant Array: \n', constant_Array)

random_Array = np.random.random((2, 3))
print('Random Array: \n', random_Array)

sequence = np.arange(0, 10, 1)
print('Sequence Array: ',sequence)
```

## Vector, Matrices and Tensor

```python
vector = np.array([1, 2, 3])
print('Vector: ',vector)

matrix = np.array([[1, 2, 3], 
                   [4, 5, 6]])
print('Matrix: \n',matrix)

tensor = np.array([[[1, 2], [3, 4]], 
                  [[5, 6], [7, 8]]])
print('Tensor: \n',tensor)
```

## Array Properties

```python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
print('Shape: ', arr.shape)
print('Dimension: ',arr.ndim)
print('Size: ',arr.size)
print('Data Type: ',arr.dtype)
```

## Array Reshaping

```python
arr = np.arange(12)
print('Original Array: ', arr)

reshaped = arr.reshape((3, 4))
print('Reshaped Array:\n',reshaped)

flattened = reshaped.flatten()
print('Flattened Reshaped Array:',flattened)

# Ravel Returrn View Instead Of Copy
raveled = reshaped.ravel()
print('Raveled Reshaped Array:',raveled)

transpose = reshaped.T 
print('Transpose Reshaped Array:\n',transpose)
```
