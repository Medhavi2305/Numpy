### Advance Operation With Business Examples

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data Structure: [restrarunt_id, 2021, 2022, 2023, 2024, 2025]
sales_data = np.array([
   [1, 150000, 200000, 250000, 300000, 350000, 600000], # Res1
   [2, 70000, 800000, 900000, 390000, 350000, 700000], # Res2
   [3, 155000, 206000, 254000, 303000, 352000, 600000], # Res3
   [4, 151000, 205000, 255000, 305000, 350600, 606000], # Res4
   [5, 150300, 202000, 250040, 300300, 350030, 667000], # Res5
   [6, 190000, 250000, 250300, 350000, 355000, 650000] # Res6
])

print('=== Sales Analtsis ===')
print('Shape Of This Data:',sales_data.shape)
print('Sample Data Of First Three Restrarunt:\n',sales_data[:3]) # This [:3] is Same as This [0:3]
print('Data Of All Restrarunts Without ID:\n',sales_data[:,1:]) # This [:,1:] Mean Start From Starting Then THis ,1 Start From First Coloumn And Then Till Ending
print('Sales Data:',np.sum(sales_data, axis=0))
print('Sales Data Where Axis Is 1:',np.sum(sales_data[:,1:], axis=1))
```

```python
yearlt_total = np.sum(sales_data[:,1:], axis=0) # Thi Is Adding The All Sales Data Not The Id 
print('Yearly Total:',yearlt_total)
```

### Understand Axis

```python
Min_Sales = np.min(sales_data[:,1:], axis=1)
print('Minimum Sales Per Year Of Each Restrarunt (→)',Min_Sales) # This Way → Left To Right
Max_Sales = np.max(sales_data[:,1:], axis=0)
print('Maximum Sales Per Year (↓)',Max_Sales) # This Way ↓ Up To Down
```

### Average 

```python
average_sales = np.mean(sales_data[:,1:], axis=1)
print('Average Sales Per Restrarunt:',average_sales)
```

### Cumilative Sum

```python
cumsum = np.cumsum(sales_data[:,1:], axis=1)
print('Cumilative Sales Per Year Of a Restrarunt:\n',cumsum)
```

### Graph

```python
plt.figure(figsize=(6,6))
plt.plot(np.mean(cumsum, axis=0))
plt.title('Average Cumulative Sales Across All Restrarut')
plt.xlabel('Years')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

### Vectors

```python
vector1 = np.array([1, 2, 3, 4, 5])
vector2 = np.array([6, 7, 8, 9, 10])
print('Vector Addition:',vector1 + vector2)
print('Vector Multiplication:',vector1 * vector2)
print('Dot Product:',np.dot(vector1,vector2))
```

### Angel In Vectors

```python
angel = np.arccos(np.dot(vector1,vector2)/(np.linalg.norm(vector1)*np.linalg.norm(vector2)))
print('Angel In Vectors:',angel)
```

### Vectorized Operations

```python
restrarunt_types = np.array(['Indian', 'French', 'Italian', 'Japanese', 'Chinese', 'Thai'])
vectorized_upper = np.vectorize(str.upper)
print('Vectorized Upper:',vectorized_upper(restrarunt_types))
```
